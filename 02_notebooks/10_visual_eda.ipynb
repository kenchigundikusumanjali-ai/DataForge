```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../01_data/processed/dataforge_cleaned.csv")
```

```python
df["Total Score"].hist()

plt.title("Total Score Distribution")

plt.savefig("../03_visuals/histogram.png")

plt.show()
```

Most students have medium total scores.

```python
plt.boxplot(df["attendance_pct"])

plt.title("Attendance Boxplot")

plt.savefig("../03_visuals/boxplot.png")

plt.show()
```

Attendance has a few extreme values.

```python
df.groupby("department")["Total Score"].mean().plot(kind="bar")

plt.title("Average Total Score by Department")

plt.savefig("../03_visuals/bar_chart.png")

plt.show()
```

Average scores differ across departments.

```python
df["Total Score"].plot()

plt.title("Total Score Trend")

plt.savefig("../03_visuals/line_chart.png")

plt.show()
```

Student scores vary across the dataset.

```python
plt.scatter(df["attendance_pct"], df["Total Score"])

plt.xlabel("Attendance")

plt.ylabel("Total Score")

plt.savefig("../03_visuals/scatter.png")

plt.show()
```

```python
Higher attendance generally relates to better scores.
```

```python
import matplotlib.pyplot as plt

corr = df.select_dtypes(include="number").corr()

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)

plt.yticks(range(len(corr.columns)), corr.columns)

plt.savefig("../03_visuals/heatmap.png")

plt.show()
```

Numeric columns show positive and negative relationships.

```python
(df.isna().sum()).plot(kind="bar")

plt.title("Missing Values")

plt.savefig("../03_visuals/missing_chart.png")

plt.show()
```

Very few missing values remain after cleaning.

```python
df.groupby("Performance")["Total Score"].mean().plot(kind="bar")

plt.title("Performance Comparison")

plt.savefig("../03_visuals/grouped_chart.png")

plt.show()
```

Good performers have higher total scores.
